```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

# 1. Dataset
X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([0, 0, 0, 1, 1, 1])

# 2. Train the model
model = LogisticRegression()
model.fit(X, y)

# 3. Extract parameters
w = model.coef_[0][0]
b = model.intercept_[0]

print(f"w = {w:.4f}")
print(f"b = {b:.4f}")

# 4. Generate x-values from 0 to 10
x_range = np.linspace(0, 10, 300)

# 5. Calculate sigmoid
z = w * x_range + b
sigmoid = 1 / (1 + np.exp(-z))

# 6. Plot
plt.figure(figsize=(8, 5))

plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(x_range, sigmoid, color='red', label='Sigmoid Curve', linewidth=2)

plt.axhline(0.5, color='gray', linestyle='--', label='Decision Boundary (0.5)')

plt.xlabel("Study Hours")
plt.ylabel("Probability of Passing")
plt.title("Logistic Regression: 0 to 10 Hours")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression

# =========================
# 1. Define the dataset
# =========================

X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([0, 0, 0, 1, 1, 1])

# =========================
# 2. Train the model
# =========================

model = LogisticRegression()
model.fit(X, y)

# =========================
# 3. Extract parameters
# =========================

w = model.coef_[0][0]
b = model.intercept_[0]

print("Logistic Regression Model")
print(f"w = {w:.4f}")
print(f"b = {b:.4f}")
print(f"f(x) = sigmoid({w:.4f} * x + {b:.4f})")

# =========================
# 4. Predict for new students
# =========================

X_test = np.array([[2], [3], [4], [5], [7]])

# Predict probabilities
probabilities = model.predict_proba(X_test)[:, 1]

# Predict classes
predictions = model.predict(X_test)

# =========================
# 5. Display results
# =========================

print("\nPrediction Results:")

for i in range(len(X_test)):
    hours = X_test[i][0]
    probability = probabilities[i]
    prediction = predictions[i]

    if prediction == 1:
        result = "Pass"
    else:
        result = "Fail"

    print(f"Study Hours: {hours}")
    print(f"Passing Probability: {probability:.2%}")
    print(f"Prediction: {result}")
    print("-------------------------")

# =========================
# 6. Plot sigmoid curve
# =========================

x_range = np.linspace(0, 8, 300)

z = w * x_range + b
sigmoid = 1 / (1 + np.exp(-z))

plt.figure(figsize=(9, 5))

plt.scatter(X, y, color="blue", label="Training Data")
plt.plot(x_range, sigmoid, color="red", label="Sigmoid Curve")

plt.scatter(
    X_test,
    probabilities,
    color="green",
    marker="x",
    s=80,
    label="Predicted Probabilities"
)

plt.axhline(
    0.5,
    color="gray",
    linestyle="--",
    label="Decision Boundary (0.5)"
)

plt.xlabel("Study Hours")
plt.ylabel("Probability of Passing")
plt.title("Logistic Regression Prediction")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LogisticRegression

# =========================
# 1. Define the dataset
# =========================

# Dachshund dogs
dach_length = [55, 57, 64, 63, 58, 49, 54, 61]
dach_height = [30, 31, 36, 30, 33, 25, 37, 34]

# Jindo dogs
jin_length = [56, 47, 56, 46, 49, 53, 52, 48]
jin_height = [52, 52, 50, 53, 50, 53, 49, 54]

# =========================
# 2. Create X and y
# =========================

X = np.array([
    [l, h]
    for l, h in zip(
        dach_length + jin_length,
        dach_height + jin_height
    )
])

# Dachshund = 0, Jindo = 1
y = np.array([0] * len(dach_length) + [1] * len(jin_length))

# Display dataset
print("--- X (Feature Data: 16x2) ---")
print(X)

print("\n--- y (Label Data: 1D Array) ---")
print(y)

# =========================
# 3. Train the model
# =========================

model = LogisticRegression()
model.fit(X, y)

print("\nModel training completed!")

# =========================
# 4. Display model parameters
# =========================

print("\nModel coefficients:", model.coef_)
print("Model intercept:", model.intercept_)

# =========================
# 5. Predict for a new dog
# =========================

# Example: length = 50, height = 52
new_dog = np.array([[50, 52]])

prediction = model.predict(new_dog)[0]
probabilities = model.predict_proba(new_dog)[0]

print("\n--- New Dog Prediction ---")
print("Length:", new_dog[0][0])
print("Height:", new_dog[0][1])

print(f"Dachshund probability: {probabilities[0]:.4f}")
print(f"Jindo probability: {probabilities[1]:.4f}")

if prediction == 0:
    print("Predicted breed: Dachshund")
else:
    print("Predicted breed: Jindo")

# =========================
# 6. Plot the dataset
# =========================

plt.figure(figsize=(8, 6))

plt.scatter(
    dach_length,
    dach_height,
    color="blue",
    label="Dachshund",
    marker="o"
)

plt.scatter(
    jin_length,
    jin_height,
    color="red",
    label="Jindo",
    marker="s"
)

plt.scatter(
    new_dog[0][0],
    new_dog[0][1],
    color="green",
    label="New Dog",
    marker="*",
    s=200
)

plt.xlabel("Length (cm)")
plt.ylabel("Height (cm)")
plt.title("Dog Breed Classification Using Logistic Regression")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# ---- Visualization ----
plt.scatter(dach_length, dach_height, c='r', label='Dachshund')
plt.scatter(jin_length, jin_height, c='b', marker='^', label='Jindo dog')

# --- (1) Predict model decision line ---
# create a range of x (length)
x_values = np.linspace(min(X[:,0])-2, max(X[:,0])+2, 200)

# decision boundary: model.coef_[0][0] * x + model.coef_[0][1] * y + intercept = 0
# solve for y: y = -(w1*x + b)/w2
w1 = model.coef_[0][0]
w2 = model.coef_[0][1]
b  = model.intercept_[0]

y_values = -(w1 * x_values + b) / w2

plt.plot(x_values, y_values, 'g--', linewidth=2, label='Decision Boundary')

plt.xlabel('Length')
plt.ylabel('Height')
plt.title("Dog Classification with Decision Line")
plt.legend(loc='upper right')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---- Visualization ----
plt.scatter(dach_length, dach_height, c='r', label='Dachshund')
plt.scatter(jin_length, jin_height, c='b', marker='^', label='Jindo dog')

# ---- Decision Boundary ----
x_values = np.linspace(min(X[:, 0]) - 2, max(X[:, 0]) + 2, 200)

w1 = model.coef_[0][0]
w2 = model.coef_[0][1]
b = model.intercept_[0]

y_values = -(w1 * x_values + b) / w2

plt.plot(
    x_values,
    y_values,
    'g--',
    linewidth=2,
    label='Decision Boundary'
)

# ---- New Data ----
newdata_length = [59]
newdata_height = [35]

newdata = np.array([[newdata_length[0], newdata_height[0]]])

# Predict the breed
y_pred = model.predict(newdata)
y_pred_prob = model.predict_proba(newdata)

# Display prediction
print("New dog length:", newdata_length[0])
print("New dog height:", newdata_height[0])

if y_pred[0] == 0:
    print("Predicted breed: Dachshund")
else:
    print("Predicted breed: Jindo dog")

print(f"Dachshund probability: {y_pred_prob[0][0] * 100:.2f}%")
print(f"Jindo dog probability: {y_pred_prob[0][1] * 100:.2f}%")

# Plot the new dog
plt.scatter(
    newdata_length,
    newdata_height,
    c='g',
    marker='p',
    s=200,
    label='New Data'
)

# ---- Labels and Title ----
plt.xlabel('Length')
plt.ylabel('Height')
plt.title('Dog Classification with Decision Line')
plt.legend(loc='upper right')
plt.grid(True)

plt.show()
```
