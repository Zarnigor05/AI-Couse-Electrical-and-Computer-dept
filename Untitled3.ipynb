```python
# Required libraries
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
import matplotlib.pyplot as plt

# 1. Create data
data = {
    'Income': [60000, 55000, 70000, 50000, 52000,
               48000, 45000, 40000, 30000, 35000,
               60000, 52000, 70000, 48000, 45000,
               50000, 51000, 49000, 53000, 55000],
    'Age':    [32, 35, 40, 30, 33,
               28, 25, 22, 27, 20,
               25, 28, 29, 30, 31,
               25, 29, 35, 31, 30],
    'Buy':    ['Yes', 'Yes', 'Yes', 'Yes', 'Yes',
               'No', 'No', 'No', 'No', 'No',
               'No', 'No', 'No', 'No', 'No',
               'No', 'No', 'No', 'Yes', 'Yes']
}

# Create DataFrame
df = pd.DataFrame(data)

# 2. Print DataFrame
print("--- DataFrame Output ---")
print(df)
```

```python
# Scatter plot
plt.figure(figsize=(8,6))

# Buy = Yes
yes = df[df['Buy'] == 'Yes']
plt.scatter(yes['Income'], yes['Age'], color='green', label='Buy: Yes', s=100, marker='o')

# Buy = No
no = df[df['Buy'] == 'No']
plt.scatter(no['Income'], no['Age'], color='red', label='Buy: No', s=100, marker='x')

# 그래프 설정
plt.xlabel('Income')
plt.ylabel('Age')
plt.title('Income vs Age Scatter Plot (Buy Yes/No)')
plt.legend()
plt.grid(True)
plt.show()
# 3. Define Features (X) and Target (y)
X = df[['Income', 'Age']]
y = df['Buy']

# 4. Train Decision Tree Model
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

# 5. Visualize the Decision Tree
plt.figure(figsize=(12, 8))
tree.plot_tree(model,
               feature_names=['Income', 'Age'],
               class_names=['No', 'Yes'],
               filled=True)
plt.title("Decision Tree for 'Buy' Prediction")
plt.show()
```

```python
# 5️. Prediction
sample = pd.DataFrame({'Income':[ 47000, 51000], 'Age':[29, 31]})
pred = model.predict(sample)
print(pred)  # ['Yes', 'No']

####  Part 5 ###
# 6. Model rule print. not function f(x)
from sklearn.tree import export_text

# 트리의 규칙을 텍스트 형태로 출력
tree_rules = export_text(model, feature_names=['Income', 'Age'])
print(tree_rules)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 4. Scatter plot과 결정 경계 시각화
plt.figure(figsize=(10, 6))

# 결정 경계를 위한 meshgrid 생성
x_min, x_max = X['Income'].min() - 5000, X['Income'].max() + 5000
y_min, y_max = X['Age'].min() - 2, X['Age'].max() + 2

xx, yy = np.meshgrid(np.arange(x_min, x_max, 500),
                     np.arange(y_min, y_max, 0.5))

# 각 점에 대해 예측
Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = np.array([1 if val == 'Yes' else 0 for val in Z])
Z = Z.reshape(xx.shape)  # reshape 오타 수정

# 결정 경계 그려주기 (Contour plot)
plt.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.coolwarm)

# 실제 데이터 산점도 (Scatter Plot) 예시 (X와 y 데이터 필요)
# plt.scatter(X['Income'], X['Age'], c=y, edgecolors='k', cmap=plt.cm.coolwarm)
plt.xlabel('Income')
plt.ylabel('Age')
plt.title('Decision Boundary')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier

# 1. Ma'lumotlar bazasi
data = {
    'Income': [60000, 55000, 70000, 50000, 52000,
               48000, 45000, 40000, 30000, 35000,
               60000, 52000, 70000, 48000, 45000,
               50000, 51000, 49000, 53000, 55000],
    'Age':    [32, 35, 40, 30, 33,
               28, 25, 22, 27, 20,
               25, 28, 29, 30, 31,
               25, 29, 35, 31, 30],
    'Buy':    ['Yes', 'Yes', 'Yes', 'Yes', 'Yes',
               'No', 'No', 'No', 'No', 'No',
               'No', 'No', 'No', 'No', 'No',
               'No', 'No', 'No', 'Yes', 'Yes']
}

df = pd.DataFrame(data)

# 2. X va y ni aniqlash hamda modelni o'qitish
X = df[['Income', 'Age']]
y = df['Buy']

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

# 3. Yangi ma'lumot (New Data)
new_income = 62000
new_age = 32

new_data = pd.DataFrame({'Income': [new_income], 'Age': [new_age]})
new_pred = model.predict(new_data)[0]

# 4. Grafik va Decision Boundary chizish
plt.figure(figsize=(10, 6))

x_min, x_max = X['Income'].min() - 5000, X['Income'].max() + 5000
y_min, y_max = X['Age'].min() - 2, X['Age'].max() + 2

xx, yy = np.meshgrid(np.arange(x_min, x_max, 500),
                     np.arange(y_min, y_max, 0.5))

grid_data = pd.DataFrame({'Income': xx.ravel(), 'Age': yy.ravel()})
Z = model.predict(grid_data)
Z = np.array([1 if val == 'Yes' else 0 for val in Z])
Z = Z.reshape(xx.shape)

# Qaror chegarasi fonini chizish (pastel pushti/qizil va ko'k tusda)
plt.contourf(xx, yy, Z, alpha=0.3, cmap=plt.cm.coolwarm)

# Asl ma'lumotlar nuqtalarini chiqarish
yes = df[df['Buy'] == 'Yes']
no = df[df['Buy'] == 'No']
plt.scatter(yes['Income'], yes['Age'], color='green', label='Buy: Yes', s=100, marker='o')
plt.scatter(no['Income'], no['Age'], color='red', label='Buy: No', s=100, marker='x')

# Yangi nuqtani ko'k yulduzcha (*) bilan belgilash (rasmdagidek)[cite: 8]
plt.scatter(new_income, new_age, color='blue', marker='*', s=250,
            label=f'New Data (Income: {new_income}, Age: {new_age} -> {new_pred})')

# Grafik bezaklari va sarlavhasi
plt.xlabel('Income')
plt.ylabel('Age')
plt.title('Decision Tree: Income vs Age with Decision Boundary & New Data')[cite: 8]
plt.legend(loc='upper left')
plt.grid(True)
plt.show()
```
